### Setting up the Environment

```python
import os
from pathlib import Path
from dotenv import load_dotenv
from azure.ai.documentintelligence import DocumentIntelligenceClient
from azure.ai.documentintelligence.models import AnalyzeDocumentRequest
import pprint

load_dotenv()

endpoint = os.getenv("FOUNDRY_AI_SERVICES_ENDPOINT")
api_version = "2024-11-30"
```

### Creating the Document Intelligence Client

```python
#creating a document analysis client
from azure.identity import DefaultAzureCredential


client = DocumentIntelligenceClient(
    endpoint=endpoint, api_version=api_version, credential=DefaultAzureCredential())
```

### Analyzing a Document from test1.jpg

```python
## test1.jpg in the same directory as this notebook
document_path = Path("test1.jpg")
document_bytes = document_path.read_bytes()
```

```python
poller = client.begin_analyze_document(
    model_id="doc-intel-custom",
    body = AnalyzeDocumentRequest(
        bytes_source=document_bytes,
    )
)
result = poller.result()

for idx, document in enumerate(result.documents):
    print("--------Analyzing document #{}--------".format(idx + 1))
    print("Document has type {}".format(document.doc_type))
    print("Document has confidence {}".format(document.confidence))
    print("Document was analyzed by model with ID {}".format(result.model_id))
    for name, field in document.fields.items():
        field_value = field.value_string if field.value_string else field.content
        print("type: '{}' name: '{}' value: '{}' confidence: {}".format(field.type, name, field_value, field.confidence))
```
